## Working on Real-world type data

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data Structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],        # Farhan Biryani
    [2, 120000, 140000, 160000, 190000],        # Ice n Spice
    [3, 200000, 230000, 260000, 300000],        # Javed Nihari
    [4, 180000, 210000, 240000, 270000],        # Kababjees
    [5, 160000, 185000, 205000, 230000],        # Balochistan Sajji
])

print("===== Foodpanda Sales Analysis =====")

print(f"\nSales data shape: {sales_data.shape}")

print(f"\nSample data for first 3 restaurants: \n{sales_data[:3]}")

print(f"\nCasual data: \n{sales_data[:, 1:]}")
# first colon (:) is for rows, since no value is given so all rows selected
# second colon (:) is for columns, starting from index 1 column till the end
```

### Sum of Sales per year

```python
sum_of_sales = np.sum(sales_data[:, 1:], axis=0)

print(f"Sum of sales per year:\n{sum_of_sales}")
```

### Minimum sales of each restaurant among all the years

```python
min_sale = np.min(sales_data[:, 1:], axis=1)

print(f"Min sales of each restaurant among all the years: \n{min_sale}")
```

### Maximum sales each year among all the restaurant

```python
max_sale = np.max(sales_data[:, 1:], axis=0)
print(f"Max sales each year among all the restaurant: \n{max_sale}")
```

### Average sales per restaurant

```python
avg_sales = np.mean(sales_data[:, 1:], axis=1)

print(f"Average sales of each restaurant: \n{avg_sales}")
```

### Cummulative Sales of each restaurant

The `cumulative sum` means the total of a sequence of numbers added step by step.

It keeps adding each number to the sum of all the previous numbers.

```python
cum_sales = np.cumsum(sales_data[:, 1:], axis=1)

print(f"Cummulative sales of each restaurant: \n{cum_sales}")
```

### Plotting a graph on Cummulative sales

```python
plt.figure(figsize=(8, 5))
plt.plot(np.mean(cum_sales, axis= 0))   # sums the sales amount of all the restaurant in one year, then take the mean.
plt.title("Average Cummulative Sales of restaurant across years")
plt.xlabel("Years")
plt.ylabel("Average Sales")
plt.grid(True)
plt.show()
```

### Vector addition, multiplication and dot product

- **Addition:** `Rows` and `columns` of both matrices/vectors should be equal.
- **Multiplication:** `Columns` of vector/matrix A should be equal to `rows` of vector/matrix B.
- **Dot Product:** `Dot product (scalar)` is a special multiplication between two vectors. Both vectors must have the same size. The result of the dot product is a `single number (scalar).` Multiply each pair of corresponding elements and then add all those products.

```python
vector_1 = np.array([1, 2, 3, 4, 5])
vector_2 = np.array([6, 7, 8, 9, 10])

print(f"Vector Sum: \n{vector_1 + vector_2}")

print(f"\nVector Multiplication: \n{vector_1 * vector_2}")

print(f"\nDot Product of two vectors: \n{np.dot(vector_1, vector_2)}")
```

### Vectorized Operation

Doing many operations on whole arrays/vectors in one step instead of looping through elements one by one. It’s faster and cleaner!

```python
restaurant_types = np.array(["chinese", "fast food", "pizza", "bbq", "biryani"])

vectorized = np.vectorize(str.upper)

print(f"Normal: {restaurant_types}")
print(f"\nVectorized to Uppercase: {vectorized(restaurant_types)}")
```

### Broadcast

**Broadcasting** in NumPy means applying operations to arrays of different shapes, where NumPy automatically expands the smaller one to match the bigger one — **without using loops**.

---

#### ✅ Example:

import numpy as np

a = np.array([1, 2, 3])  # 1D array
b = 5                    # scalar value

result = a + b
print(result) ---> [6, 7, 8]
